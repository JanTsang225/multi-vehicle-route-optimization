```python
import pandas as pd
import numpy as np
```

```python
file_path1 = r"E:\SQL learning 2\airbnb_project\Bristol\download\csv_file\Bristol_calendar.csv"
calendar = pd.read_csv(file_path1, encoding="latin1")  # use latin1 to avoid UnicodeDecodeError

file_path2 = r"E:\SQL learning 2\airbnb_project\Bristol\download\csv_file\Bristol_listings.csv"
listings = pd.read_csv(file_path2, encoding="latin1")  # use latin1 to avoid UnicodeDecodeError
```

```python
calendar_ids = set(calendar["listing_id"].unique())
listing_ids = set(listings["id"].unique())

print("Calendar only:", len(calendar_ids - listing_ids))
print("Listings only:", len(listing_ids - calendar_ids))
print("Overlap:", len(calendar_ids & listing_ids))
```

```python
print(listings.columns.tolist())
print(listings.columns.tolist()[0])
```

```python
total_rows = len(listings)
unique_ids = len(listings["id"].unique())
print(f"Total rows: {total_rows}, Unique IDs: {unique_ids}")
```

```python
duplicates = listings[listings["id"].duplicated(keep=False)]  # keep=False shows all occurrences
print(duplicates)  # or print(duplicates.head()) to see first few
```

```python
grouped = duplicates.groupby("id")
for id_val, group in grouped:
    if len(group.drop_duplicates()) > 1:  # more than one unique row
        print(f"Differences for ID {id_val}:")
        print(group)
        print("-" * 50)
```
